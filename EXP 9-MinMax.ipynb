```python
import math

# Board cells: empty ' ', player 'X', opponent 'O'
board = [' ' for _ in range(9)]

def print_board(b):
    for i in range(3):
        print(f" {b[3*i]} | {b[3*i+1]} | {b[3*i+2]} ")
        if i < 2:
            print("-----------")

def is_winner(b, player):
    wins = [
        (0, 1, 2), (3, 4, 5), (6, 7, 8),  # rows
        (0, 3, 6), (1, 4, 7), (2, 5, 8),  # columns
        (0, 4, 8), (2, 4, 6)             # diagonals
    ]
    for a, b_, c in wins:
        if b[a] == b[b_] == b[c] == player:
            return True
    return False

def is_board_full(b):
    return ' ' not in b

def minimax(b, depth, is_maximizing):
    # Terminal States: Base cases
    if is_winner(b, 'X'):
        return 10 - depth  # Maximizer ('X') wins
    if is_winner(b, 'O'):
        return depth - 10  # Minimizer ('O') wins
    if is_board_full(b):
        return 0  # Draw

    if is_maximizing:
        best_score = -math.inf
        for i in range(9):
            if b[i] == ' ':
                b[i] = 'X'  # Try move
                score = minimax(b, depth + 1, False) #Recurse for minimizer
                b[i] = ' '  # Undo move
                best_score = max(best_score, score)
        return best_score # Return after checking all moves
    else: # Minimizing player
        best_score = math.inf
        for i in range(9):
            if b[i] == ' ':
                b[i] = 'O' # Try move
                score = minimax(b, depth + 1, True) # Recurse for maximizer
                b[i] = ' ' # Undo move
                best_score = min(best_score, score)
        return best_score # Return after checking all moves

def best_move(): # Moved outside minimax
    """
    Find the best move for maximizer (X)
    """
    best_score = -math.inf
    move = -1
    for i in range(9):
        if board[i] == ' ': # Corrected typo from -- to ==
            board[i] = 'X'
            score = minimax(board, 0, False)
            board[i] = ' '
            if score > best_score:
                best_score = score
                move = i
    return move

def play_game(): # Moved outside minimax
    """
    Simple game loop: Human (O) vs. AI (X using Minimax)
    """
    print("Board positions are numbered 0 to 8:") # Corrected typo from Borard to Board
    print("0 | 1 | 2")
    print("-----------")
    print("3 | 4 | 5")
    print("-----------")
    print("6 | 7 | 8")
    print()

    while True:
        print_board(board)
        # Check if game ended before AI's turn
        if is_winner(board, 'O'):
            print("You Win!")
            break
        if is_board_full(board):
            print("Game Over! It's a Tie.")
            break

        # AI move
        move = best_move()
        if move != -1:
            board[move] = 'X'
            print(f"AI (X) played at position {move}")
        else:
            # This case should ideally be caught by is_board_full earlier
            print("Game Over! It's a Tie.")
            break

        print_board(board)
        if is_winner(board, 'X'):
            print("AI (X) Wins!")
            break
        if is_board_full(board):
            print("Game Over! It's a Tie.")
            break

        # Human Move
        while True:
            try:
                human_move = int(input("Enter your move (0-8): "))
                if 0 <= human_move < 9 and board[human_move] == ' ':
                    board[human_move] = 'O'
                    break
                else:
                    print("Invalid move. Try again.")
            except ValueError:
                print("Invalid input. Enter a number.")

if __name__ == "__main__":
    play_game()
```
